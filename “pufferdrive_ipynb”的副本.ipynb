```python
import os
import shutil
from pathlib import Path
import glob
```

```python
Path("data/processed/training").mkdir(parents=True, exist_ok=True)
Path("resources/drive/binaries").mkdir(parents=True, exist_ok=True)
```

```python
possible_paths = [
    "GPUDrive_mini/train/*.json",
    "GPUDrive_mini/training/*.json",
    "GPUDrive_mini/**/*.json"
]

source_files = []
for pattern in possible_paths:
    files = glob.glob(pattern, recursive=True)
    if files:
        source_files = files
        print(f"Found {len(files)} files in {pattern}")
        break

if not source_files:
    print("No JSON files found. Let's check the structure:")
    !ls -R GPUDrive_mini/ | head -50
else:
    for src_file in source_files:
        dst_file = os.path.join("data/processed/training", os.path.basename(src_file))
        shutil.copy2(src_file, dst_file)

    num_copied = len(glob.glob("data/processed/training/*.json"))
    print(f"✓ Copied {num_copied} JSON files to data/processed/training/")
```

```python
import shutil
import os

binary_dir = "resources/drive/binaries"
if os.path.exists(binary_dir):
    shutil.rmtree(binary_dir)
    print("✓ Removed old binary files")
os.makedirs(binary_dir, exist_ok=True)
print("✓ Created fresh binary directory")
```

```python
import time
import sys

import numpy as np
print(f"NumPy version: {np.__version__}")
print("\nConverting JSON maps to binary format...")
start_time = time.time()

from pufferlib.ocean.drive.drive import process_all_maps
process_all_maps()

elapsed = time.time() - start_time
print(f"\n✓ Conversion completed in {elapsed/60:.2f} minutes")

# Verify
import glob
binary_files = glob.glob("resources/drive/binaries/*.bin")
print(f"✓ Created {len(binary_files)} binary map files")
```
